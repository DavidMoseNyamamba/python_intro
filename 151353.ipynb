```python
def euclidean_distance(x1, x2):
    """ Calculate the Euclidean distance between two points"""
    """ x1 and x2 are two numpy arrays"""
    """ return the Euclidean distance between x1 and x2"""
    return np.sqrt(np.sum((x1 - x2) ** 2))

def kNearestNeighbor(data, predict, k=3):
    """predict the class of a new data point based on the k nearest neighbors

       parameters:
         data(list): a list of lists, where each list is a data point
         predict(list): a list of the new data point
         k(int): the number of neighbors to consider

         returns:
            int: the predicted class of the new data point
    """
    predictions = []
    for row in data:
        # calculate the Euclidean distance between the new data point and each data point in the data set
        distance = euclidean_distance(row[:-1], predict)
        # append the class label and distance to the predictions list
        predictions.append((row[-1], distance))

    # sort the predictions list by distance
    predictions = sorted(predictions, key=lambda x: x[1])

    # select the k nearest neighbors
    top_k = predictions[:k]

    # count the number of each class in the top k
    result = []
    for item in top_k:  # item is a tuple
        result.append(item[0])  # append the class label to the result list

    # return the most common class label
    return max(set(result), key=result.count)

data = np.array([
    [1, 4, 'A', 'D', 'Good'],
    [5, 9, 'E', 'I', 'Good'],
    [6, 7, 'F', 'G', 'Good'],
    [1, 1, 'A', 'A', 'Good'],
    [3, 8, 'C', 'H', 'Good'],
    [10, 4, 'J', 'D', 'Bad'],
    [5, 3, 'E', 'C', 'Bad'],
    [2, 3, 'B', 'C', 'Bad'],
    [6, 3, 'F', 'C', 'Bad'],
    [2, 8, 'B', 'H', 'Bad']
])

# convert categorical data to numerical data
label_encoders = {}
for i in range(2, 4):
    le = LabelEncoder()
    data[:, i] = le.fit_transform(data[:, i])
    label_encoders[i] = le

# separate features and labels
X = data[:, :-1]
y = data[:, -1]

# new data point to classify
new_data = np.array([13, 7, 26, 25])  # x1=13, x2=7, x3=Z, x4=Z (Encoded as 26 and 25)

# predict the class of the new data point
prediction = kNearestNeighbor(data, new_data, k=3)
print(f'The new data point belongs to class: {prediction}')

# decode the predicted class back to the original label
predicted_label = label_encoders[3].inverse_transform([prediction])[0]
print(f'The new data point belongs to class: {predicted_label}')
```
